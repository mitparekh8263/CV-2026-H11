np.resize works in 3 steps - flattening -> size adjustment -> reshaping

def mock_np_resize(a, new_shape):

    # 1. Convert input to a standard array and flatten it to 1D
    a = np.asarray(a).flatten()
    
    # 2. Calculate total elements needed for the new shape
    new_size = np.prod(new_shape)
    
    # 3. Truncate if too big, or tile/repeat the data if too small
    if new_size <= a.size:
        a = a[:new_size]
    else:
        repeats = int(np.ceil(new_size / a.size))
        a = np.tile(a, repeats)[:new_size]
        
    # 4. Reshape the 1D data into the target dimensions
    return a.reshape(new_shape)

for ex: there is a 2*2 image - [[A, B]; [C, D]] and each element of matrix contains rgb value [Ri, Gi, Bi] where i = {a, b, c, d} and reqired image must be of 3*1

step 1: flattening 2D array into 1D array --> [A, B, C, D]
so in terms of rgb --> [Ra, Ga, Ba, Rb, Gb, Bb, Rc, Gc, Bc, Rd, Gd, Bd]

step 2: size adjustment --> 3*1 image required means 3 rows and there are total 12 rgb elements => each row contains 12/3 = 4 rgb elements sequentially

step 3: reshaping --> convert to required dimension matrix
=> [[Ra, Ga, Ba, Rb]; [Gb, Bb, Rc, Gc]; [Bc, Rd, Gd, Bd]]

```python
# method 2 - using cv resizing
cv_resize = cv2.resize(img, (275,182)) # no need to mention BGR channels in just width * height format
cv2_imshow(cv_resize)
cv_resize.shape
```

here we can see so much of difference b/w np.resize and cv.resize because cv understands the picture so it can blends the neighboring pixels together and scales the visual content smoothly while npr.resize consider this as long string of numbers and wraps it into matrix/grid
